# CFD Analysis Walkthrough

This notebook reads the generated metric file and explains drag, lift, and residual trends. Run `python scripts/setup_case.py` before opening this notebook.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

metrics_path = Path('../data/processed/metrics.csv')
data = pd.read_csv(metrics_path)
data.head()
```

```python
fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(data['iteration'], data['drag_coefficient'], label='Drag coefficient')
ax.plot(data['iteration'], data['lift_coefficient'], label='Lift coefficient')
ax.set_xlabel('Iteration')
ax.set_ylabel('Coefficient')
ax.set_title('Educational aerodynamic coefficient history')
ax.grid(True, alpha=0.3)
ax.legend();
```

The sample coefficients are generated for learning. In a real OpenFOAM case, similar plots can be created from force coefficient output files.
